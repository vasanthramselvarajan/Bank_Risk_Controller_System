```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import ast
```

```python
df = pd.read_csv("tmdb_5000_movies.csv")
df.head()
```

```python
data = pd.read_csv("tmdb_5000_credits.csv")
data.head()
```

```python
movie_df = pd.merge(df,data, left_on=['id',"title"], right_on=['movie_id',"title"]).drop(['movie_id'], axis=1)
movie_df.head()
```

```python
movie_df.shape
```

```python
movie_df.isnull().sum()
```

```python
movie_df = movie_df[["id","title","overview","genres","keywords","cast","crew"]].copy()
```

```python
movie_df.head()
```

```python
movie_df.info()
```

```python
movie_df.isnull().sum()
```

```python
movie_df.dropna(subset=['overview'], inplace=True)
```

```python
movie_df.iloc[0]["genres"]
```

```python
ast.literal_eval(movie_df.iloc[0]["genres"])
```

```python
def convert(obj):
    genres=[]
    for i in ast.literal_eval(obj):
        genres.append(i["name"].lower())
    return genres
```

```python
movie_df["genres"] = movie_df["genres"].apply(convert)
movie_df["keywords"] = movie_df["keywords"].apply(convert)
```

```python
movie_df[["genres","keywords"]]
```

```python
ast.literal_eval(movie_df["cast"][0])
```

```python
def cast(obj):
    cast=[]
    count = 0
    for i in ast.literal_eval(obj):
        if count < 5 :
            cast.append(i["name"].lower())
            count = count + 1
        else:
            break
    return cast
```

```python
movie_df["cast"] = movie_df["cast"].apply(cast)
```

```python
movie_df["cast"][0]
```

```python
ast.literal_eval(movie_df["crew"][0])
```

```python
def director(obj):
    director = []
    for i in ast.literal_eval(obj):
        if i["job"]=='Director':
            director.append(i["name"].lower())
    return director
```

```python
movie_df["director"] = movie_df["crew"].apply(director)
```

```python
movie_df["director"][0]
```

```python
movie_df.drop(["crew"],axis=1,inplace=True)
```

```python
movie_df.head()
```

```python
movie_df["genres_x"] = movie_df["genres"].apply(lambda x: [i.replace(" ","") for i in x])
movie_df["keywords"] = movie_df["keywords"].apply(lambda x: [i.replace(" ","") for i in x])
movie_df["cast"] = movie_df["cast"].apply(lambda x: [i.replace(" ","") for i in x])
movie_df["director_x"] = movie_df["director"].apply(lambda x: [i.replace(" ","") for i in x])
```

```python
movie_df['tags'] = movie_df["genres_x"] + movie_df["keywords"] + movie_df["cast"] + movie_df["director_x"]
```

```python
movie_df.drop(columns=["genres_x","keywords","cast","director_x"], inplace=True)
```

```python
movie_df.head(3)
```

```python
movie_df["tags"] = movie_df["tags"].apply(lambda x: " ".join(x))
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

tfv = TfidfVectorizer(stop_words='english')
vectors = tfv.fit_transform(movie_df['tags']).toarray()
```

```python
from sklearn.metrics.pairwise import cosine_similarity

similarity = cosine_similarity(vectors)
```

```python
similarity.shape
```

```python
list(enumerate(similarity[0]))
```

```python
sorted(list(enumerate(similarity[0])), reverse=True, key=lambda vec: vec[1])
```

```python
def recommend(movie_title,num_recommendations):
    # Check if the movie exists in the DataFrame
    if movie_title not in movie_df['title'].values:
        print(f"Movie '{movie_title}' not found in the database.")
        return

    # Find the index of the movie in the DataFrame
    index = movie_df[movie_df['title'] == movie_title].index[0]

    # Get the genre of the specified movie
    movie_genre = movie_df.iloc[index]['genres']

    # Calculate the similarity distances
    distance = list(enumerate(similarity[index]))
    movie_list = sorted(distance, reverse=True, key=lambda vec: vec[1])

    # Print the title and genre of the specified movie
    print(f"Recommendations for '{movie_title}' , Genre is: {movie_genre} \n") # 

    # Print the titles and genres of the top 5 similar movies
    count = 1
    for i in movie_list[1:num_recommendations+1]:  # skipping the first result as it's the same movie
        title = movie_df.iloc[i[0]]['title']
        genre = movie_df.iloc[i[0]]['genres']
        overview = movie_df.iloc[i[0]]['overview']
        print(f" Movie {count}: {title} , Genre: {genre},\n Overview: {overview}\n")
        count += 1
```

```python
# Use recommend function with "movie title" and "number of recommendations"

movie_name = input("Enter movie name: ")
num = int(input("Enter number of recommendations: "))

recommend(movie_name,num)
```
