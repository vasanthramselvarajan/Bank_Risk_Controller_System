doing chiSquare test , mutual info and feature importance by RF 

```python
# Chi2 test
from sklearn.feature_selection import chi2
from sklearn.preprocessing import LabelEncoder
from sklearn.feature_selection import SelectKBest

X_encoded = df_objects.apply(LabelEncoder().fit_transform)
y_encoded = LabelEncoder().fit_transform(df_sample['TARGET'])

selector = SelectKBest(score_func=chi2, k=15)
selector.fit_transform(X_encoded, y_encoded)

selected_columns = X_encoded.columns[selector.get_support()].to_list()[:15]
print(selected_columns)
```

```python
# Calculate Mutual Information for categorical features
from sklearn.metrics import mutual_info_score

mi_scores = []
for feature in df_objects.columns:
    mi = mutual_info_score(df_objects[feature], df_sample['TARGET'])
    mi_scores.append(mi)
```

```python
mi_series = pd.Series(mi_scores, index=df_objects.columns)
ranked_categorical_features = mi_series.sort_values(ascending=False).index.tolist()

selected_features = ranked_categorical_features[:15]

print("Selected Features:", selected_features)
```

```python
# Feature importance using RF classifier
from sklearn.ensemble import RandomForestClassifier

rf = RandomForestClassifier(random_state=42)
rf.fit(X_encoded, y_encoded)

importances = rf.feature_importances_
rf_feature_importance = pd.Series(importances, index=X_encoded.columns)
rf_top_features = rf_feature_importance.sort_values(ascending=False).head(15).index.tolist()
```

```python
rf_top_features
```

```python
# Intersection: features selected by all
intersection_obj = set(selected_columns) & set(rf_top_features) & set(selected_features)

# Union: features selected by any
union_obj = set(selected_columns) | set(rf_top_features) | set(selected_features)

print("Common Important Features (intersection):")
print(sorted(intersection_obj))

print("\nAll Selected Features (union):")
print(sorted(union_obj))
```

#### numeric columns null values imputing

```python
df_num = df_sample.select_dtypes(include=['float64', 'int64']).copy()
```

```python
for i in df_num.columns:
    if df_num[i].isnull().sum() > 0:
        print(f'{i} , null % : {(df_num[i].isnull().sum()/len(df_sample))*100:.3f} , unique value :{df_num[i].nunique()}')
```

some numeric columns are cateorical in disguice , so we are using mean , median , KNN imputer.

```python
impute_check = pd.DataFrame(index=df_num.columns)
impute_check['Null %'] = df_num.isnull().mean() * 100
impute_check['Unique'] = df_num.nunique()
impute_check['Skewness'] = df_num.skew()

impute_check = impute_check[impute_check['Null %'] > 0]

def decide_imputation(row):
    if row['Null %'] < 15:
        if abs(row['Skewness']) < 0.5 and row['Unique'] > 100:
            return 'Mean'
        elif row["Unique"]< 30 :
            return 'Mode'
        else:
            return 'Median'
    elif row['Null %'] < 50:
        return 'KNN or Median'
    else:
        return 'Consider Dropping or KNN'

impute_check['Suggested Imputation'] = impute_check.apply(decide_imputation, axis=1)

impute_check.sort_values(by='Null %', ascending=False, inplace=True)
```

```python
# Droping high null columns
df_num.drop(['LANDAREA_MODE', 'LANDAREA_AVG', 'LANDAREA_MEDI', 'BASEMENTAREA_AVG',
       'BASEMENTAREA_MEDI', 'BASEMENTAREA_MODE', 'EXT_SOURCE_1',
       'NONLIVINGAREA_MODE', 'NONLIVINGAREA_MEDI', 'NONLIVINGAREA_AVG', 'RATE_DOWN_PAYMENT', 'ELEVATORS_AVG',
       'ELEVATORS_MODE', 'ELEVATORS_MEDI', 'APARTMENTS_MEDI', 'APARTMENTS_AVG',
       'APARTMENTS_MODE'],axis=1,inplace=True)
```

```python
# Mode imputation for numeric columns
mode = ['AMT_REQ_CREDIT_BUREAU_MON', 'AMT_REQ_CREDIT_BUREAU_DAY',
       'AMT_REQ_CREDIT_BUREAU_WEEK', 'AMT_REQ_CREDIT_BUREAU_HOUR',
       'AMT_REQ_CREDIT_BUREAU_YEAR', 'AMT_REQ_CREDIT_BUREAU_QRT',
       'DEF_60_CNT_SOCIAL_CIRCLE', 'DEF_30_CNT_SOCIAL_CIRCLE',
       'OBS_60_CNT_SOCIAL_CIRCLE', 'OBS_30_CNT_SOCIAL_CIRCLE']

for i in mode:
       df_num[i].fillna(df_num[i].mode()[0],inplace=True)
```

```python
# Median imputation
for i in df_num.columns:
       if df_num[i].isnull().sum() > 0 and i not in mode:
              df_num[i].fillna(df_num[i].median(),inplace=True)
```

```python
# correlation with target variable (top 15)
correlation = df_num.corr()['TARGET'].sort_values(ascending=True).index.to_list()[:15]
print(correlation)
```

```python
# feature selection using selectkBest (top 15)
from sklearn.feature_selection import SelectKBest, f_classif
X = df_num.drop(["TARGET"],axis=1)
y = df_num["TARGET"]

selector = SelectKBest(score_func=f_classif, k=15)
X_new = selector.fit_transform(X, y)

selected_columns = X.columns[selector.get_support()].to_list()[:15]
print(selected_columns)
```

```python
# Point-biserial correlation (top 15)

from scipy.stats import pointbiserialr

result = {}
for i in df_num.columns:
    corr, p_value = pointbiserialr(df_num[i], df_num['TARGET'])
    result[i] = corr

corr_series = pd.Series(result).sort_values(ascending=False).index.to_list()[:15]
print(corr_series)
```
