## Feature Engineering

```python
# making negative values into abs() and converted into years
df["AGE"] = (df["DAYS_BIRTH"].abs()/365).round()
df["EMPLOYED_YEARS"] = (df["DAYS_EMPLOYED"].abs()/365).round()
df["LAST_PHONE_CHANGE_YEARS"] = (df["DAYS_LAST_PHONE_CHANGE"].abs()/365).round()
df["YEARS_REGISTRATION"] = (df["DAYS_REGISTRATION"].abs()/365).round()
```

```python
# Implementing "INCOME_TO_CREDIT" score
df["INCOME_TO_CREDIT"]=df["AMT_INCOME_TOTAL"]/df["AMT_ANNUITY"]
```

```python
df["AMT_INCOME_TOTAL"].quantile([0.25,0.5,0.75,0.90,0.95,0.99])
```

```python
df["AGE_CATEGORY"] = df["AGE"].apply(
    lambda age: "Young Aged" if age < 25 else ("Middle Aged" if age > 45 else "Senior Aged")
)
df["INCOME_CATEGORY"] = df["AMT_INCOME_TOTAL"].apply(
    lambda income: "Low Income" if income < 50000 else ("Middle Income" if income < 150000 else "High Income")
)
```

```python
# Removing unwanted columns
df.drop(["DAYS_BIRTH","DAYS_EMPLOYED","DAYS_LAST_PHONE_CHANGE","DAYS_REGISTRATION"],axis=1,inplace=True)
```

```python
for i in ["AGE","EMPLOYED_YEARS","LAST_PHONE_CHANGE_YEARS","YEARS_REGISTRATION"]:
    df[i]=df[i].astype(int)
```

```python
outlier_column = ['AMT_ANNUITY','AMT_CREDIT','AMT_GOODS_PRICE','AMT_INCOME_TOTAL','EXT_SOURCE_3','EMPLOYED_YEARS',
       'LAST_PHONE_CHANGE_YEARS', 'YEARS_REGISTRATION', 'INCOME_TO_CREDIT']
```

```python
for i in outlier_column:
    q1 = df[i].quantile(0.25)
    q3 = df[i].quantile(0.75)
    IQR = q3 - q1

    lower_bound = q1 - 1.5 * IQR
    upper_bound = q3 + 1.5 * IQR

    df = df[(df[i] >= lower_bound) & (df[i] <= upper_bound)]
```

## Model Training

```python
values = df.drop('TARGET', axis=1)
target = df['TARGET']
```

```python
from imblearn.over_sampling import SMOTE

sampler = SMOTE(random_state=46)
values_res, target_res = sampler.fit_resample(values, target)

print(f"Original training set shape: {target.value_counts()}")
print(f"Resampled training set shape: {target_res.value_counts()}")
```

```python
from sklearn.model_selection import train_test_split

train_data, test_data, train_lab, test_lab = train_test_split(values_res, target_res, test_size=0.2, random_state=42)
```

```python
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, confusion_matrix, classification_report, roc_auc_score, roc_curve, auc

def get_metrics(test_lab, lab_pred, model, test_data):
    print("Test Scores:")
    score=pd.DataFrame({
    "Accuracy":[accuracy_score(test_lab, lab_pred)*100],
    "Precision":[precision_score(test_lab, lab_pred)*100],
    "Recall": [recall_score(test_lab, lab_pred)*100],
    "F1 Score:": [f1_score(test_lab, lab_pred)*100],
    "ROC-AUC score":[roc_auc_score(test_lab,model.predict_proba(test_data)[:, 1])*100]
    })
    return score

def get_confusion_matrix(test_lab,lab_pred):
    cm = confusion_matrix(test_lab,lab_pred)

    print("Confusion Matrix:")

    plt.figure(figsize=(6, 4))
    sns.heatmap(cm, annot=True, fmt="d", cmap="Blues", xticklabels=["Non-Defaulter(0)","Defaulter(1)"], yticklabels=["Non-Defaulter(0)","Defaulter(1)"])
    plt.title("Confusion Matrix")
    plt.xlabel("Actual")
    plt.ylabel("Predicted")
    plt.show()

def roc_auc_curve(test_lab, model, test_data):
    false, true, thresholds = roc_curve(test_lab, model.predict_proba(test_data)[:, 1])
    roc_auc = auc(false, true)
    plt.plot(false, true, label='ROC curve (area = %0.2f)' % roc_auc)
    plt.plot([0, 1], [0, 1], 'k--')
    plt.xlim([0.0, 1.0])
    plt.ylim([0.0, 1.05])
    plt.xlabel('False Positive Rate')
    plt.ylabel('True Positive Rate')
    plt.title('Receiver operating characteristic')
    plt.legend(loc="lower right")
    plt.show()
```

### Hyper parameter tuning

```python
from sklearn.model_selection import GridSearchCV

param_grid = {
    'n_estimators': [100, 200],
    'max_depth': [10, 20],
    'min_samples_split': [2, 5, 10],
    'min_samples_leaf': [1, 2, 4],
    'bootstrap': [True]
}

rf = RandomForestClassifier(random_state=42)

# Perform Grid Search
grid_search = GridSearchCV(estimator=rf, param_grid=param_grid, cv=3, scoring='roc_auc', verbose=2, n_jobs=-1)
grid_search.fit(train_data, train_lab)

# Best parameters and score
print("Best Parameters:", grid_search.best_params_)
print("Best ROC-AUC Score:", grid_search.best_score_)

# Use the best model
best_rf = grid_search.best_estimator_
```
